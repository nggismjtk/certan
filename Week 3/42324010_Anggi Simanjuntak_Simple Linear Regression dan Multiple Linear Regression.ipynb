```python
import pandas as pd
df = pd.read_csv('Salary_Data.csv')
print(df.head())
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, color = "red")
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
# Fitting Simple Linear Regression to the Training set
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# memprediksi hasil pada data uji (Test set) menggunakan model regresi yang telah dilatih
y_pred = regressor.predict(X_test)
y_pred
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear
# Semakin mendekati 1, semakin baik model

from sklearn.metrics import r2_score
score = r2_score(y_test, y_pred)
score
```

```python
# Visualizing the Training set results
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
```

```python
# Visualizing the Test set results
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

Tugas Simple Linear Regression

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
# Membaca dataset dari file CSV di folder
df = pd.read_csv('rounded_hours_student_scores.csv')
```

```python
# Menyiapkan X dan y
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
# Membagi dataset menjadi training dan testing (80:20)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
# Fitting model
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
# Menampilkan hasil intercept, slope, dan persamaan
a = regressor.intercept_
b = regressor.coef_[0]
print("--- Simple Linear Regression ---")
print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# Memprediksi dan menghitung R-Square
y_pred = regressor.predict(X_test)
score = r2_score(y_test, y_pred)
print("R-Square Score:", score)
```

Tugas Multiple Linear Regression

```python
# Membaca dataset dari file CSV di folder
df2 = pd.read_csv('Marketing_Data.csv')
```

```python
# Menyiapkan X2 (fitur: youtube, facebook, newspaper) dan y2 (target: sales)
X2 = df2.iloc[:, :-1]
y2 = df2.iloc[:, -1]
```

```python
# Membagi dataset menjadi training dan testing (80:20)
X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size = 0.2, random_state = 0)
```

```python
# Fitting model
regressor2 = LinearRegression()
regressor2.fit(X_train2, y_train2)
```

```python
# Menampilkan hasil intercept, koefisien, dan persamaan
a_ = regressor2.intercept_
b1 = regressor2.coef_[0]
b2 = regressor2.coef_[1]
b3 = regressor2.coef_[2]

print("--- Multiple Linear Regression ---")
print("Estimated model intercept, a:", a_)
print("Estimated model slope, b1 (youtube):", b1)
print("Estimated model slope, b2 (facebook):", b2)
print("Estimated model slope, b3 (newspaper):", b3)
print(f"ŷ = {a_:.2f} + {b1:.2f}x1 + {b2:.2f}x2 + {b3:.2f}x3")
```

```python
# Memprediksi dan menghitung R-Square
y_pred2 = regressor2.predict(X_test2)
score2 = r2_score(y_test2, y_pred2)
print("R-Square Score 2:", score2)
```
