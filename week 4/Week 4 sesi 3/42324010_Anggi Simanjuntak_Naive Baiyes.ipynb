```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# load dataset Iris dari scikit-learn
iris = load_iris()
```

NumPy digunakan untuk perhitungan numerik, sedangkan Pandas digunakan untuk mengolah data dalam bentuk tabel. Selanjutnya, load_iris() dari Scikit-learn digunakan untuk mengambil dataset Iris, kemudian disimpan ke dalam variabel iris

```python
#konversi data menjadi DataFrame pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

#definisikan X (fitur) dan y (target/Label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

pd.DataFrame() membuat tabel dari data fitur Iris, kemudian df["Species"] menambahkan kolom Species yang berisi nama jenis bunga berdasarkan target dataset. selanjutnya, X berisi fitur atau data input yang digunakan untuk melakukan prediksi, sedangkan y berisi target atau label berupa jenis bunga Iris

```python
y
```

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y =lab.fit_transform(y)
y
```

untuk mengubah label jenis bunga Iris yang berupa teks menjadi angka agar dapat diproses oleh model machine learning. 
LabelEncoder() membuat objek untuk melakukan proses tersebut, kemudian lab.fit_transform(y) mengubah setiap nama spesies menjadi kode angka

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

untuk membuat dan melatih model Naive Bayes dengan algoritma GaussianNB. model = GaussianNB() membuat model Naive Bayes yang cocok untuk data numerik seperti dataset Iris. Kemudian, model.fit(X_train, y_train) melatih model menggunakan data training. kemudian hasil model yang sudah dilatih disimpan dalam variabel NB_model

```python
result = NB_model.predict(X_test)
result
```

untuk menguji model yang sudah dilatih dengan menggunakan data X_test. NB_model.predict(X_test) meminta model naive bayes untuk memprediksi jenis bunga berdasarkan data fitur pada X_test

```python
from sklearn.metrics import accuracy_score

print('Accuracy of data testing: ', accuracy_score(y_test, result) * 100)
```

accuracy_score(y_test, result) membandingkan label sebenarnya (y_test) dengan hasil prediksi model (result). Hasilnya kemudian dikalikan 100 agar ditampilkan dalam bentuk persentase. jadi, print() akan menampilkan persentase data testing yang berhasil diprediksi dengan benar oleh model.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

untuk memprediksi jenis bunga iris berdasarkan data baru. input_data berisi nilai empat fitur bunga. Selanjutnya, NB_model.predict(input_data) digunakan untuk memprediksi kelas bunga berdasarkan data tersebut. Hasil prediksi yang masih angka kemudian diubah menjadi nama spesies menggunakan lab.inverse_transform(result)

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

untuk memprediksi jenis bunga Iris dari data baru. pd.DataFrame() mengubah data [5.1, 3.5, 1.4, 0.2] menjadi tabel dengan nama kolom sesuai fitur pada dataset iris
